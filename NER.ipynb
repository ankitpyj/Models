```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = "Elon Musk founded SpaceX in the United States in 2002."

doc = nlp(text)
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
 #Visualize entities
from spacy import displacy
displacy.render(doc, style="ent")
```

ANOTHER EXAMPLE

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = """
The Indian Space Research Organisation is the national
space agency of India, headquartered in Bengaluru.
"""

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_, spacy.explain(ent.label_))
```

```python
from spacy import displacy
displacy.render(doc, style="ent")
```
